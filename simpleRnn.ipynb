```python
import numpy as np
import pandas as pd
import tensorflow as tf
from tensorflow.keras.datasets import imdb
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, SimpleRNN, Dense, Input
from tensorflow.keras.preprocessing import sequence
```

```python
# load imdb dataset
max_features = 10000 # vocabulary size
(x_train, y_train), (x_test, y_test) = imdb.load_data(num_words=max_features)

## print shape of data
print('Training data shape' , x_train.shape)
print('Training label shape',y_train.shape)
print('Testing data shape', x_test.shape)
print('Testing label shape', y_test.shape)
```

```python
## inspect sample review and its label
sample_review = x_train[0]
sample_label = y_train[0]
print('Sample review:', sample_review)
print('Sample label:', sample_label)
```

```python
## mapping of word index back to words

word_index = imdb.get_word_index()
print('word index', word_index)
reverse_word_index = {value: key for (key, value) in word_index.items()}
print('reverse word index', reverse_word_index)
```

```python
decoded_review = ' '.join([reverse_word_index.get(i - 3, '?') for i in sample_review])
print(decoded_review)
```

```python
max_length = 500
x_train = sequence.pad_sequences(x_train, maxlen=max_length)
x_test = sequence.pad_sequences(x_test, maxlen=max_length)
print('x_train shape:', x_train.shape)
print('x_test shape:', x_test.shape)
```

```python
## train model with simpleRNN

model = Sequential()
model.add(Input(shape=(max_length,)))
model.add(Embedding(max_features, 128))
model.add(SimpleRNN(128,activation='relu'))
model.add(Dense(1, activation='sigmoid'))

```

```python
model.summary()
```

```python
## create a instance of earlystopping callback

from tensorflow.keras.callbacks import EarlyStopping
early_stopping = EarlyStopping(monitor='val_loss', patience=5, restore_best_weights=True)
early_stopping
```

```python
# compile the model

model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])
```

```python
## train the model with earlystopping

history= model.fit(
    x_train, y_train,
    batch_size=32,
    epochs=10,
    validation_split=0.2,
    callbacks=[early_stopping]
)
```

```python
model.save('simple_rnn_imdb.h5')
```
